## 1. Import Required Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## 2. Load the Dataset

```python
df = pd.read_csv("C:\\PROGRAMS\\lungcancer.csv")
df.head()
```

## 3. Check Dataset Shape

```python
print("Dataset Shape:", df.shape)
```

## 4. Inspect Data Types and Non-Null Counts

```python
df.info()
```

## 5. Summary Statistics

```python
df.describe()
```

## 6. Drop Unnecessary Columns

```python
df.drop(columns=["index", "Patient Id"], errors="ignore", inplace=True)
df.head()
```

## 7. Rename Misspelled Columns

```python
df.rename(columns={
    "OccuPational Hazards": "Occupational Hazards",
    "chronic Lung Disease": "Chronic Lung Disease"
}, inplace=True)
df.columns
```

## 8. Convert Target Variable to Categorical

```python
df["Level"] = df["Level"].astype("category")
df['Level'].dtype
```

## 9. Handle Missing Values in Shortness of Breath

```python
df["Shortness of Breath"] = df["Shortness of Breath"].fillna(df["Shortness of Breath"].mean())
df.isnull().sum()
```

## 10. Check Cancer Risk Level Distribution

```python
df["Level"].value_counts()
```

## 11. Average Age by Risk Level

```python
df.groupby("Level", observed=True)["Age"].mean()
```

## 12. Average Smoking Score by Risk Level

```python
df.groupby("Level", observed=True)["Smoking"].mean()
```

## 13. Average Air Pollution Score by Risk Level

```python
df.groupby("Level", observed=True)["Air Pollution"].mean()
```

## 14. Visualization 1: Cancer Level Distribution (Pie Chart)

```python
plt.figure(figsize=(6, 6))
df["Level"].value_counts().plot.pie(autopct='%1.1f%%', startangle=90, colors=sns.color_palette('pastel'))
plt.title('Distribution of Cancer Risk Levels')
plt.ylabel('')
plt.show()
```

## 15. Visualization 2: Average Age by Risk Level (Bar Chart)

```python
plt.figure(figsize=(8, 5))
df.groupby("Level", observed=True)["Age"].mean().plot(kind='bar', color='skyblue')
plt.title('Average Age by Risk Level')
plt.xlabel('Risk Level')
plt.ylabel('Average Age')
plt.xticks(rotation=0)
plt.show()
```

## 16. Visualization 3: Age Distribution across Risk Levels (Box Plot)

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x="Level", y="Age", data=df, palette="Set2")
plt.title('Age Distribution Across Risk Levels')
plt.show()
```

## 17. Visualization 4: Age vs Air Pollution (Scatter Plot)

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x="Age", y="Air Pollution", hue="Level", data=df, palette="deep")
plt.title('Age vs Air Pollution by Risk Level')
plt.show()
```

Correlation Heatmap (excluding non-numeric/ID columns)

```python
# Correlation Heatmap (excluding non-numeric/ID columns)
plt.figure(figsize=(12, 10))
numeric_df = df.drop(columns=['Patient Id', 'index'])
# Encode Level for correlation analysis if desired
numeric_df['Level_Encoded'] = numeric_df['Level'].map({'Low': 1, 'Medium': 2, 'High': 3})
numeric_df = numeric_df.drop(columns=['Level'])
```

Heatmap: Correlation between numerical columns

```python
# Heatmap: Correlation between numerical columns
plt.figure(figsize=(12, 8))

sns.heatmap(
    df.select_dtypes(include="number").corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
risk_counts = df['Level'].value_counts()
plt.figure(figsize=(7, 7))

plt.pie(
    risk_counts,
    labels=risk_counts.index,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops={'width': 0.4},
    pctdistance=0.8,
    colors=['#ff9999', '#66b3ff', '#99ff99']
)

plt.title('Lung Cancer Risk Level Distribution')
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=df,
    x='Age',
    y='Air Pollution',
    hue='Level',
    hue_order=['Low', 'Medium', 'High'],
    palette={'Low': 'green', 'Medium': 'orange', 'High': 'red'},
    alpha=0.7
)

plt.title('Age vs Air Pollution by Risk Level')
plt.xlabel('Patient Age')
plt.ylabel('Air Pollution Score')
plt.legend(title='Risk Level')
plt.show()
```

```python
df['Age_Group'] = pd.cut(df['Age'], bins=[0, 20, 30, 40, 50, 60, 80], labels=['<20', '20-30', '31-40', '41-50', '51-60', '60+'])
age_pollution = df.groupby('Age_Group', observed=False)['Air Pollution'].mean()

plt.figure(figsize=(10, 5))
plt.plot(age_pollution.index.astype(str), age_pollution.values, marker='o', linestyle='-', color='teal')

plt.title('Average Air Pollution Score by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Average Air Pollution Score')
plt.grid(True)
plt.show()
```
